```python
import requests

import requests
import os
import mimetypes

def speech_to_text(file_path, language_code="hi-IN", model="saarika:v1", with_timestamps=True):
    url = "https://api.sarvam.ai/speech-to-text"

    # Check if file exists
    if not os.path.exists(file_path):
        return f"Error: File '{file_path}' not found."

    # Check file type
    file_type, _ = mimetypes.guess_type(file_path)
    allowed_types = ['audio/mpeg', 'audio/wave', 'audio/wav', 'audio/x-wav']
    if file_type not in allowed_types:
        return f"Error: File type '{file_type}' not allowed. Allowed types are {allowed_types}"

    # Prepare the multipart form data
    try:
        files = {
            'file': ('audio_file', open(file_path, 'rb'), file_type)
        }
    except IOError as e:
        return f"Error opening file: {str(e)}"

    data = {
        'language_code': language_code,
        'model': model,
        'with_timestamps': str(with_timestamps).lower()
    }

    headers = {
        "api-subscription-key": "6159e8a9-baff-4fad-a897-edff6a686c29"
    }

    # Send the POST request
    try:
        response = requests.post(url, files=files, data=data, headers=headers)
        response.raise_for_status()  # Raises an HTTPError for bad responses
        return response.json()
    except requests.exceptions.RequestException as e:
        return f"Error during API request: {str(e)}"
    finally:
        files['file'][1].close()  # Ensure the file is closed

# Example usage
result = speech_to_text("output.wav")
print(result['transcript'])

# Example usage
```
